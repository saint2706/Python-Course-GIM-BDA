**The single most important distinction:** `print()` *shows* a value on screen and returns nothing; `return` *hands the value back* so you can store it, add to it, or feed it to another function. A function that prints instead of returning will fail almost every exam question that asks you to "write a function that returns…".

**Default arguments** (`def apply_gst(amount, rate=0.18)`) let the caller skip an argument. Defaults must come *after* all non-default arguments.

**Q6. Write a function `avg_rating(*ratings)` that returns the average of any number of numeric ratings??**

````python
# Solution for Q6
def avg_rating(*ratings: float) -> float:
    """Return the arithmetic mean of the supplied rating values."""
    if not ratings:
        raise ValueError('At least one rating is required')
    return sum(ratings) / len(ratings)

print(round(avg_rating(4.2, 3.8, 4.5, 4.0), 2))
````

**Q7. Write a function `format_invoice(**fields)` that returns a string like `INV:<id> - <client> - <amount>` using keys `id`, `client`, `amount` from kwargs??**

**Q8. Write a function `percent_change(old, new)` that returns percentage change rounded to 2 decimals (positive or negative)??**

````python
# Solution for Q8
def percent_change(old: float, new: float) -> float:
    """Return the percentage change from old to new rounded to two decimals."""
    if old == 0:
        raise ValueError('Old value must be non-zero for percentage change')
    change = ((new - old) / old) * 100
    return round(change, 2)

print(percent_change(100, 120))
````

### ⚡ Beyond the syllabus: docstrings and type hints: free marks, zero risk

A docstring (the triple-quoted text just under `def`) and type hints (`price: float -> float`) cost one line each and make the function self-documenting. Python doesn't enforce the hints, so they can never break your code, but they tell the reader (and the examiner) exactly what goes in and what comes out. `help(fn)` then works on your own functions.

````python
def apply_gst(amount: float, rate: float = 0.18) -> float:
    """Return `amount` with GST added.

    Args:
        amount: pre-tax value in rupees.
        rate:   GST rate as a decimal. Defaults to 0.18 (18%).

    Returns:
        The tax-inclusive amount, rounded to 2 decimals.

    Examples:
        >>> apply_gst(1000)
        1180.0
    """
    return round(amount * (1 + rate), 2)

print(apply_gst(1000))            # uses the default 18%
print(apply_gst(1000, 0.05))      # override for a 5% slab
help(apply_gst)                   # your docstring becomes the help page

# Self-testing with assert — three lines that prove the function works.
# If any assertion is False the cell stops with an error, so silence = success.
assert apply_gst(1000) == 1180.0
assert apply_gst(0) == 0.0
assert apply_gst(1000, 0) == 1000.0
print("\nAll checks passed — the function behaves as documented.")
````

## Part 3: Functions that work on collections (Q11-Q15)

These take a list or dictionary in and give a list or dictionary back. Q11 ("top n sellers") is the exam favourite, and it's `sorted(..., key=..., reverse=True)[:n]` wrapped in a `def`.

**Q11. Write a function `top_sellers(sales_dict, n)` that takes a dict like `{'A':1200,'B':900,'C':1500}` and returns a list of the top `n` product IDs by sales (highest first)??**

````python
# Solution for Q11
def top_sellers(sales_dict: dict[str, float], n: int) -> list[str]:
    """Return the top n product IDs sorted by sales in descending order."""
    sorted_items = sorted(sales_dict.items(), key=lambda item: item[1], reverse=True)
    return [product for product, _ in sorted_items[:n]]

print(top_sellers({'A': 1200, 'B': 900, 'C': 1500}, 2))
````

**Q12. Write a function `filter_high_value(orders, threshold)` where `orders` is a list of amounts; return only amounts strictly above `threshold`??**

````python
# Solution for Q12
def filter_high_value(orders: list[float], threshold: float) -> list[float]:
    """Return orders that are strictly above the provided threshold."""
    return [order for order in orders if order > threshold]

print(filter_high_value([1200, 800, 4500], 1000))
````

**Q13. Write a function `tag_discounts(prices, cutoff)` that returns `['HIGH' if p>=cutoff else 'LOW' for each price]` without using list comprehensions (use a loop)??**

````python
# Solution for Q13
def tag_discounts(prices: list[float], cutoff: float) -> list[str]:
    """Return HIGH/LOW tags for each price using an explicit loop."""
    tags = []
    for price in prices:
        if price >= cutoff:
            tags.append('HIGH')
        else:
            tags.append('LOW')
    return tags

print(tag_discounts([120, 250, 90, 300], 200))
````

**Q14. Write a function `merge_stocks(main, incoming)` that merges two dictionaries of stocks by summing counts for matching keys and keeping non-overlapping ones??**

````python
# Solution for Q14
def merge_stocks(main: dict[str, int], incoming: dict[str, int]) -> dict[str, int]:
    """Merge two stock dictionaries, summing counts for matching product codes."""
    combined = main.copy()
    for product, qty in incoming.items():
        combined[product] = combined.get(product, 0) + qty
    return combined

print(merge_stocks({'P01': 50, 'P02': 20}, {'P02': 30, 'P03': 15}))
````

````python
# --- Another valid way to write the same answer -------------------
# WHY show two? Examiners give credit for either. Pick whichever you
# can reproduce from memory under time pressure.
def merge_stocks(main, incoming):
    merged = main.copy()  # Start with main dictionary
    for key, value in incoming.items():
        if key in merged:
            merged[key] += value  # Sum counts for matching keys
        else:
            merged[key] = value   # Add non-overlapping keys
    return merged

# Example usage
main_stocks = {'A': 100, 'B': 200, 'C': 50}
incoming_stocks = {'B': 150, 'C': 30, 'D': 70}

result = merge_stocks(main_stocks, incoming_stocks)
print(result)
````

## Part 4: Recursion, closures and safety (Q16-Q20)

**Recursion** is a function that calls itself. Every recursive function needs two parts: a **base case** that stops it (`if n == 0: return 1`) and a **recursive step** that moves toward the base case. Miss the base case and it runs until Python gives up.

**A closure** (Q18) is a function that *returns another function*, remembering the values around it. **`try/except`** (Q19) catches an error instead of letting it crash the notebook, which is worth real marks in a data-cleaning question.

**Q16. Write a recursive function `factorial(n)` that returns n! and raises `ValueError` if `n` is negative??**

````python
# Solution for Q16
def factorial(n: int) -> int:
    """Return n! using recursion, raising ValueError for negative inputs."""
    if n < 0:
        raise ValueError('Factorial is not defined for negative numbers')
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

**Q17. Write a recursive function `flatten(nested)` that flattens a list like `[[1,2],[3,[4]]]` into `[1,2,3,4]`??**

````python
# Solution for Q17
def flatten(nested: list) -> list:
    """Recursively flatten a nested list structure."""
    result: list = []
    for item in nested:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

print(flatten([[1, 2], [3, [4]]]))
````

````python
# --- Another valid way to write the same answer -------------------
# WHY show two? Examiners give credit for either. Pick whichever you
# can reproduce from memory under time pressure.
def flatten(nested):
    flat_list = []
    for item in nested:
        if isinstance(item, list):
            flat_list.extend(flatten(item))  # Recursively flatten sublist
        else:
            flat_list.append(item)           # Append non-list item
    return flat_list

# Example usage
nested_list = [[1, 2], [3, [4]]]
result = flatten(nested_list)
print(result)
````

**Q18. Write a function `make_incrementer(step)` that returns a new function which, when called with `x`, returns `x + step` (closure)??**

````python
# Solution for Q18
def make_incrementer(step: int):
    """Return a closure that increments numbers by the configured step."""
    def increment(value: int) -> int:
        return value + step
    return increment

add_five = make_incrementer(5)
print(add_five(10))
````

**Q19. Write a function `safe_div(a, b)` that returns `a/b` but returns `None` if `b==0` instead of raising an error??**

````python
# Solution for Q19
def safe_div(a: float, b: float) -> float | None:
    """Return a/b if possible; otherwise return None when divisor is zero."""
    if b == 0:
        return None
    return a / b

print(safe_div(10, 2))
print(safe_div(10, 0))
````

### ⚡ Beyond the syllabus: decorators and caching: wrapping a function without editing it

A decorator is a function that wraps another one to add behaviour (timing, logging, caching) without touching the original code. `@lru_cache` is the ready-made one worth knowing: it memorises results, turning an exponential recursive function into an instant one. Showing this on the recursive question is exactly the kind of thing that earns 'innovative method' credit.

````python
import time
from functools import lru_cache, wraps

# ---- 1. A decorator you write yourself: time any function -------------------
def timed(fn):
    @wraps(fn)                        # keeps the original name and docstring
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = fn(*args, **kwargs)
        print(f"   {fn.__name__}({args[0] if args else ''}) took {(time.perf_counter()-start)*1000:8.2f} ms")
        return result
    return wrapper

# ---- 2. The classic slow recursion ------------------------------------------
def fib_slow(n):
    return n if n < 2 else fib_slow(n - 1) + fib_slow(n - 2)

# ---- 3. The same function, with results cached ------------------------------
@lru_cache(maxsize=None)              # <- one line; remembers every result
def fib_fast(n):
    return n if n < 2 else fib_fast(n - 1) + fib_fast(n - 2)

timed(fib_slow)(28)
timed(fib_fast)(28)
print("\nSame answer:", fib_slow(28) == fib_fast(28))
print("Cache stats:", fib_fast.cache_info())
print("\n🔧 CHANGE THIS: put @lru_cache above ANY pure function that gets called")
print("   repeatedly with the same arguments — recursive or not.")
````

### ⚡ Beyond the syllabus: error handling that a data pipeline can survive

Q19's `safe_div` is the seed of a much bigger idea. Real data has blanks, text where numbers should be, and division by zero. A function that *validates its input and reports what went wrong* is worth far more than one that crashes on row 4,318 of the CSV.

````python
def safe_div(a, b, default=None):
    """Divide a by b; return `default` instead of raising when it can't."""
    try:
        return a / b
    except ZeroDivisionError:
        print(f"   warning: tried to divide {a} by zero -> returning {default}")
        return default
    except TypeError:
        print(f"   warning: {a!r} / {b!r} isn't numeric -> returning {default}")
        return default

print("normal   :", safe_div(10, 4))
print("by zero  :", safe_div(10, 0))
print("bad type :", safe_div(10, 'x'))

# Cleaning a messy column — the real-world version
raw_values = ['1200', '900', 'N/A', '', '1500', 'unknown', '2100.50']

def to_number(text, default=0.0):
    """Convert text to a float, substituting `default` when it isn't a number."""
    try:
        return float(text)
    except (ValueError, TypeError):
        return default

cleaned = [to_number(v) for v in raw_values]
rejected = [v for v in raw_values if to_number(v, None) is None]
print(f"\ncleaned : {cleaned}")
print(f"rejected: {rejected}   <- always report what you dropped, never discard silently")

# raise: refuse bad input loudly when a silent default would hide a real problem
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError(f"factorial is undefined for negative numbers (got {n})")
    return 1 if n <= 1 else n * factorial(n - 1)

print("\nfactorial(5) =", factorial(5))
try:
    factorial(-3)
except ValueError as e:
    print("caught as intended:", e)
````

## Exam quick-reference

| To do this | Write this |
|---|---|
| Define / call | `def f(a, b):` … `f(1, 2)` |
| Hand a value back | `return value` |
| Default argument | `def f(x, rate=0.18):` |
| Any number of arguments | `def f(*values):` |
| Named extras | `def f(**options):` |
| Docstring | `"""One line saying what it does."""` |
| Type hints | `def f(x: float) -> float:` |
| Recursion | base case + call to itself |
| Function returning a function | `def outer(s): return lambda x: x + s` |
| Catch an error | `try: … except ValueError: …` |
| Raise your own | `raise ValueError('message')` |
| Cache repeat calls | `@lru_cache(maxsize=None)` |
| Quick self-test | `assert f(2) == 4` |
| Thousands separator | `f'{amount:,.2f}'` |

### Adapting this in the exam

- Question says 'returns' → use `return`, not `print`. Then `print(f(...))` to show the result.
- Question says 'any number of' → `*args`.
- Question says 'optional' or 'defaults to' → a default argument.
- Question says 'without using the built-in X' → write the loop version by hand (see Q10).
- Question says 'handle the error' → `try/except`, and say in a comment what you return instead.

### Traps that cost marks

- **`print` is not `return`.** A function that prints returns `None`, so `total = my_func(...)` silently stores nothing.
- Never use a mutable default (`def f(items=[])`). The list is created once and shared between calls. Use `def f(items=None)` then `items = items or []`.
- Defaults must come after non-defaults: `def f(a=1, b)` is a syntax error.
- A recursive function with no base case hits `RecursionError`. Write the stopping condition first.
- Variables created inside a function are invisible outside it. Return them if you need them.
- Bare `except:` swallows every error including your typos. Always name the exception you expect.
