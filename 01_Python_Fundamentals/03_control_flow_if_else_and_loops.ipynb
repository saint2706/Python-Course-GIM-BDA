## Part 1: `if` / `elif` / `else`

Three rules and you know the whole of it:

1. The line ends with a **colon**, and the block underneath is **indented** (4 spaces). Indentation *is* the grouping in Python; get it wrong and the logic changes.
2. `elif` is checked only if everything above it was False. Order matters: put the most specific test first.
3. `else` is the catch-all and takes no condition.

Chain comparisons naturally: `30 <= age <= 50` is valid Python and reads exactly like the business rule.

```python
# Solution for Q1
monthly_sales = 120000
target = 100000
if monthly_sales > target:  # evaluate target achievement
    print('Target Achieved')
```

### ⚡ Beyond the syllabus: `match` / `case`: the clean way to write Q8-style branching

Python 3.10 added `match`, the tidy replacement for a long `if/elif/elif/else` chain that compares one variable against several fixed values. It's clearer to read, harder to get wrong, and shows you're current with the language.

```python
client_type = 'Premium'

# THE FAMILIAR WAY
if client_type == 'Premium':
    benefits = 'Free shipping + 20% off + priority support'
elif client_type == 'Standard':
    benefits = 'Free shipping over ₹999'
elif client_type == 'Basic':
    benefits = 'Standard rates apply'
else:
    benefits = 'Unknown tier'

# THE MODERN WAY — same logic, one variable named once
match client_type:
    case 'Premium':
        benefits_v2 = 'Free shipping + 20% off + priority support'
    case 'Standard':
        benefits_v2 = 'Free shipping over ₹999'
    case 'Basic':
        benefits_v2 = 'Standard rates apply'
    case _:                       # `_` is the catch-all, same job as `else`
        benefits_v2 = 'Unknown tier'

print(benefits)
print(benefits_v2)
print("Identical result:", benefits == benefits_v2)

# BONUS: match can test several values at once
for tier in ['Premium', 'Gold', 'Basic', 'Trial']:
    match tier:
        case 'Premium' | 'Gold':      # | means "or"
            print(f"{tier:<8} -> loyalty desk")
        case 'Basic' | 'Standard':
            print(f"{tier:<8} -> self-service")
        case _:
            print(f"{tier:<8} -> onboarding team")
```

## Part 2: `for` loops

`for item in collection:` means "give me each item, one at a time, and call it `item`". The name is yours to choose; pick a meaningful one (`for order in orders:` beats `for x in orders:`).

Three variants worth memorising:
- `for i in range(1, 6):` gives the numbers 1,2,3,4,5 (stops **before** 6)
- `for k, v in d.items():` gives a dictionary's keys and values together
- `for i, item in enumerate(lst):` gives position **and** value at the same time

**Q11. How can you iterate over a list of departments `['HR','Finance','Sales']` and print each??**

```python
# Solution for Q11
departments = ['HR', 'Finance', 'Sales']
for department in departments:
    print(f"Department: {department}")
```

## Part 3: `while` loops

Use `while` when you don't know in advance how many rounds you need, "keep deducting stock until it hits zero", "keep doubling until we pass ₹10,000".

**The one rule:** something inside the loop must eventually make the condition False, or it runs forever. If a cell hangs, that's the reason: press ■ (stop) and check that your counter actually changes.

**Q16. How can you use while loop to keep reducing stock=10 until it reaches 0??**

```python
# Solution for Q16
stock = 10
while stock > 0:
    print(f"Stock remaining: {stock}")
    stock -= 1  # decrement each loop
print('Stock depleted')
```

**Q17. How can you use while loop with flag=True to process tasks until task_count=5??**

```python
# Solution for Q17
flag = True
task_count = 0
while flag:
    print(f"Processing task #{task_count + 1}")
    task_count += 1
    flag = task_count < 5  # stop after five iterations
```

**Q18. How can you simulate while loop where balance=100 keeps reducing by 20 until balance <=0??**

```python
# Solution for Q18
balance = 100
while balance > 0:
    print(f"Balance before deduction: {balance}")
    balance -= 20
print('Balance exhausted')
```

**Q19. How can you use while loop with user input simulation `n=3` to print countdown??**

```python
# Solution for Q19
n = 3
while n > 0:
    print(n)
    n -= 1
print('Countdown complete')
```

**Q20. How can you use while loop with flag to print 'Running' three times and then stop??**

```python
# Solution for Q20
flag = True
count = 0
while flag:
    print('Running')
    count += 1
    if count == 3:
        flag = False  # break condition
```

## Part 4: Loops with decisions inside

This is the pattern behind most real questions and most exam marks: walk the list, test each item, do something with the ones that match. Three flavours:

- **Count**: start at 0, `+= 1` when the test passes
- **Collect**: start with `[]`, `.append()` when the test passes
- **Find the first**: test inside the loop, then `break` the moment you succeed

**Q21. How can you check if sales list `[120,250,90]` has values above 200 and count them using for loop??**

```python
# Solution for Q21
sales_values = [120, 250, 90]
high_count = 0
for value in sales_values:
    if value > 200:
        high_count += 1
print(f"Number of high sales entries: {high_count}")
```

### ⚡ Beyond the syllabus: `for ... else` and `break`: the 'search' idiom

Q27 asks for the *first* negative number. Most people set a `found = False` flag. Python has a built-in way: attach an `else` to the `for` loop. It runs **only if the loop finished without hitting `break`**, i.e. only if nothing was found. It removes the flag variable entirely.

```python
readings = [10, -5, 20, -8]

for value in readings:
    if value < 0:
        print(f"First negative found: {value}")
        break                     # stop immediately — no need to check the rest
else:
    # runs ONLY if the loop never hit `break`
    print("No negative values at all — all readings healthy.")

# The same search on data that is genuinely clean
for value in [10, 20, 30]:
    if value < 0:
        print("negative:", value)
        break
else:
    print("No negative values at all — all readings healthy.")

# `next()` with a generator does it in one line (returns a default if nothing matches)
first_negative = next((v for v in readings if v < 0), None)
print("One-liner version:", first_negative)
```

### ⚡ Beyond the syllabus: skip the loop entirely: vectorised thinking

This is the single biggest idea separating a beginner from an analyst. A loop touches one value at a time; **pandas and NumPy touch the whole column at once** and run 10-100x faster. Every classify/count/filter loop above has a one-line vectorised twin. Learn to see the translation: from Section 02 onward you'll be expected to write the right-hand column.

```python
import numpy as np
import pandas as pd

marks = [45, 65, 85, 32, 91, 58]

# ---- THE LOOP (Q23) -------------------------------------------------------
grades_loop = []
for m in marks:
    if m >= 80:
        grades_loop.append('Excellent')
    elif m >= 60:
        grades_loop.append('Good')
    else:
        grades_loop.append('Pass')

# ---- THE VECTORISED TWIN --------------------------------------------------
# np.select takes a list of conditions and a matching list of answers.
# It checks them IN ORDER and takes the first that is True — exactly like if/elif.
marks_arr = np.array(marks)
conditions = [marks_arr >= 80, marks_arr >= 60]      # 🔧 CHANGE THIS: your cut-offs
choices    = ['Excellent',      'Good']              # 🔧 CHANGE THIS: your labels
grades_vec = np.select(conditions, choices, default='Pass')   # default = the final `else`

print("loop      :", grades_loop)
print("vectorised:", list(grades_vec))
print("identical :", grades_loop == list(grades_vec))

# ---- The same three counting patterns, without loops -----------------------
s = pd.Series(marks)
print("\nCount above 60 :", (s > 60).sum())        # True counts as 1
print("Collect above 60:", s[s > 60].tolist())      # boolean filtering
print("Sum of all      :", s.sum())
print("Only evens      :", s[s % 2 == 0].tolist())
```

## Exam quick-reference

| To do this | Write this |
|---|---|
| Single decision | `if sales > target:` |
| Two-way decision | `if …: … else: …` |
| Three or more | `if …: elif …: else: …` |
| Between two values | `if 30 <= age <= 50:` |
| Both must be true | `if a > 0 and b > 0:` |
| Either is enough | `if a > 0 or b > 0:` |
| Loop a list | `for dept in departments:` |
| Loop numbers 1-5 | `for i in range(1, 6):` |
| Loop a dictionary | `for k, v in d.items():` |
| Position **and** value | `for i, x in enumerate(lst):` |
| Repeat until false | `while balance > 0:` |
| Leave the loop early | `break` |
| Skip to the next item | `continue` |
| Running total | `total += value` |
| Count matches | `if cond: count += 1` |

### Adapting this in the exam

- Different threshold? Only the number after the comparison operator changes.
- Asked to *collect* instead of *count*? Swap `count += 1` for `results.append(item)`.
- Asked for the *last* match instead of the first? Drop the `break` and keep overwriting the variable.
- Asked to do it 'without a loop'? Reach for `np.select`, boolean filtering or a comprehension.

### Traps that cost marks

- `=` assigns, `==` compares. `if x = 5:` is a syntax error, the exam's most common typo.
- Forgetting the colon at the end of `if` / `for` / `while`, or mixing tabs and spaces in the indented block.
- A `while` loop whose counter never changes runs forever. Always update the variable inside the loop.
- `range(1, 5)` gives 1,2,3,4, **not** 5. To include the end, add one.
- Changing a list while looping over it skips items. Loop over a copy instead, with `for x in lst[:]:`, or build a new list.
- `elif` order matters: with `if m >= 60` before `if m >= 80`, nothing is ever 'Excellent'. Test the strictest condition first.
