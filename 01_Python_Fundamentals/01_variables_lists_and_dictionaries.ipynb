## Part 1: Numbers, operators and printing

A *variable* is just a name you stick on a value so you can reuse it. The operators are the ones from school maths, plus two worth memorising: `%` gives the **remainder** after division (useful for 'does this divide evenly?' questions) and `**` means **to the power of** (useful for growth forecasts).

Notice every answer below ends with a `print(f"...")`. That `f` lets you drop a live value straight into a sentence: in an exam, a labelled answer reads far better than a bare number.

**Q1. How can you calculate the total cost if a product price is 250 and shipping charge is 40??**

```python
# Solution for Q1
product_price = 250  # base selling price of the product
shipping_charge = 40  # flat logistics fee
total_cost = product_price + shipping_charge  # combine fixed and variable components
print(f"Total payable amount: ₹{total_cost}")
```

**Q2. How can you find the remainder when total sales 1234 are divided equally among 12 salespeople??**

```python
# Solution for Q2
total_sales = 1234  # overall revenue collected
salespeople = 12  # number of staff to split among
remainder = total_sales % salespeople  # modulus captures the leftover amount
print(f"Remainder after equal split: {remainder}")
```

**Q3. What is the Python expression to calculate 7 raised to the power of 4 for forecasting growth??**

```python
# Solution for Q3
base = 7
exponent = 4
forecasted_growth = base ** exponent  # exponentiation operator models repeated multiplication
print(f"7 raised to the power of 4 = {forecasted_growth}")
```

### ⚡ Beyond the syllabus: build a formatted report, not a wall of prints

Examiners reward output that a manager could actually read. Three tools do the heavy lifting: **format specifiers** inside f-strings (`:,.2f` = thousands separator + 2 decimals), **`:>10`** to right-align a column, and a loop to print rows. Same numbers, far better presentation.

```python
# WHAT: a tiny report generator. Same data, presentation-ready output.
branch_revenue = {'Delhi': 8500000, 'Mumbai': 12400000, 'Bengaluru': 9730500}
total = sum(branch_revenue.values())

print(f"{'BRANCH':<12}{'REVENUE (₹)':>16}{'SHARE':>9}")   # <12 = left-align in 12 chars, >16 = right-align
print("-" * 37)
for branch, revenue in branch_revenue.items():
    share = revenue / total * 100
    #  :>16,.0f  ->  right-aligned, thousands separators, no decimals
    print(f"{branch:<12}{revenue:>16,.0f}{share:>8.1f}%")
print("-" * 37)
print(f"{'TOTAL':<12}{total:>16,.0f}{100:>8.1f}%")
```

## Part 2: Lists, the workhorse

A list is an ordered sequence written in square brackets. Two rules trip everyone up:

- **Counting starts at 0.** `clients[0]` is the *first* client, `clients[2]` is the *third*.
- **Slices stop one short.** `inventory[2:5]` gives positions 2, 3 and 4, never 5. Read the colon as "up to, but not including".

Negative numbers count backwards from the end, so `orders[-1]` is always the last item however long the list is.

**Q11. Given the list of clients `clients=['Tata','Infosys','Reliance','Adani']`, how can you access the second element??**

```python
# Solution for Q11
clients = ['Tata', 'Infosys', 'Reliance', 'Adani']
second_client = clients[1]  # zero-based indexing
print(f"Second client in the list: {second_client}")
```

**Q12. How can you get the last order from the list `orders=['ORD1','ORD2','ORD3','ORD4']` using negative indexing??**

```python
# Solution for Q12
orders = ['ORD1', 'ORD2', 'ORD3', 'ORD4']
last_order = orders[-1]  # negative index counts from the end
print(f"Most recent order: {last_order}")
```

**Q13. How can you retrieve items from index 2 to 4 in the list `inventory=[100,200,300,400,500]`??**

```python
# Solution for Q13
inventory = [100, 200, 300, 400, 500]
subset = inventory[2:5]  # slice from index 2 up to but not including 5
print(f"Items at positions 2-4: {subset}")
```

## Part 3: Dictionaries, labelled data

A dictionary stores **key → value** pairs. Instead of remembering that position 3 holds the location, you just ask for `supplier['location']`. This is exactly how a pandas row works, so getting comfortable here pays off twice.

The safe way to read a key is `supplier.get('rating', 'not rated')`: if the key is missing you get your fallback instead of a crash. Under exam pressure that one habit prevents a lot of red error text.

**Q21. How can you create a dictionary for a supplier with keys 'name','location' and values 'ABC Ltd','Mumbai'??**

```python
# Solution for Q21
supplier = {'name': 'ABC Ltd', 'location': 'Mumbai'}
print(f"Supplier record: {supplier}")
```

**Q22. How can you access the 'location' value from dictionary `supplier={'name':'ABC Ltd','location':'Mumbai'}`??**

```python
# Solution for Q22
supplier = {'name': 'ABC Ltd', 'location': 'Mumbai'}
location = supplier['location']  # direct key lookup
print(f"Supplier location: {location}")
```

**Q23. How can you add a new key 'rating' with value 4.5 in dictionary `supplier={'name':'ABC Ltd','location':'Mumbai'}`??**

```python
# Solution for Q23
supplier = {'name': 'ABC Ltd', 'location': 'Mumbai'}
supplier['rating'] = 4.5  # assign a new key-value pair
print(f"Supplier with rating: {supplier}")
```

**Q24. How can you update the value of 'location' to 'Delhi' in dictionary `supplier={'name':'ABC Ltd','location':'Mumbai'}`??**

```python
# Solution for Q24
supplier = {'name': 'ABC Ltd', 'location': 'Mumbai'}
supplier['location'] = 'Delhi'  # overwrite existing key value
print(f"Updated supplier: {supplier}")
```

**Q28. How can you check if key 'rating' exists in dictionary `supplier={'name':'ABC Ltd','location':'Mumbai'}`??**

**Q30. How can you create a dictionary mapping product codes `['P01','P02']` to stock counts `[150,300]`??**

### ⚡ Beyond the syllabus: build a dictionary from two lists with `zip` and a dict comprehension

Q30 asks you to pair product codes with stock counts. The manual way is a loop; the professional way is one line. `zip` walks two lists side by side like a zipper, and a **dict comprehension** `{k: v for k, v in ...}` builds the dictionary as it goes. This exact pattern reappears constantly in pandas (`dict(zip(df.code, df.stock))`).

```python
codes  = ['P01', 'P02', 'P03', 'P04']
stock  = [150, 300, 0, 42]

# THE ONE-LINER: zip pairs them up, the {k: v for ...} builds the dictionary
stock_by_code = {code: qty for code, qty in zip(codes, stock)}
print("Built in one line:", stock_by_code)

# Same idea, filtered — only products that need reordering
reorder = {code: qty for code, qty in zip(codes, stock) if qty < 100}
print("Needs reordering:  ", reorder)

# Flip it around: value -> key (handy for lookups in the other direction)
code_by_stock = {qty: code for code, qty in stock_by_code.items()}
print("Reversed mapping:  ", code_by_stock)
```

### ⚡ Beyond the syllabus: counting things properly with `Counter` and `defaultdict`

"How many orders per city?" is the single most common data question there is. Writing it with a plain dictionary takes five lines and an `if key not in dict` check. `Counter` does it in one and hands you `.most_common()` for free, this is the pure-Python ancestor of `df.groupby().size()`.

```python
cities = ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Delhi', 'Chennai']

# THE LONG WAY (what most people write)
counts_manual = {}
for c in cities:
    counts_manual[c] = counts_manual.get(c, 0) + 1   # .get(c, 0) avoids a KeyError
print("Manual   :", counts_manual)

# THE SHORT WAY
print("Counter  :", dict(Counter(cities)))
print("Top 2    :", Counter(cities).most_common(2))   # sorted for you, highest first

# GROUPING (not just counting) — defaultdict never raises KeyError
orders = [('Delhi', 1200), ('Mumbai', 900), ('Delhi', 450)]
by_city = defaultdict(list)          # any new key starts life as an empty list
for city, amount in orders:
    by_city[city].append(amount)     # no "if city not in by_city" needed
print("Grouped  :", dict(by_city))
print("City totals:", {c: sum(v) for c, v in by_city.items()})
```

## Exam quick-reference

| To do this | Write this |
|---|---|
| Add / multiply / power | `a + b` · `a * b` · `a ** b` |
| Remainder (divisibility test) | `360 % 12 == 0` |
| Round to nearest whole number | `round(4578.678)` |
| Average of a list | `sum(vals) / len(vals)` |
| Swap two variables | `a, b = b, a` |
| First / last item | `lst[0]` · `lst[-1]` |
| Items 3rd to 5th | `lst[2:5]` (stops before 5) |
| Add / remove from a list | `lst.append(x)` · `lst.remove(x)` |
| Reverse a list | `lst[::-1]` (copy) or `lst.reverse()` (in place) |
| Is it in the list? | `'Infosys' in clients` |
| Numbers 101…110 | `list(range(101, 111))` |
| Read a key safely | `d.get('rating', 'N/A')` |
| Add / update / delete a key | `d['k'] = v` · `del d['k']` |
| All keys / values / pairs | `d.keys()` · `d.values()` · `d.items()` |
| Loop keys **and** values | `for k, v in d.items():` |
| Two lists → dictionary | `dict(zip(keys, values))` |

### Adapting this in the exam

- Different numbers in the question? Only the values on the right of the `=` change. The structure never does.
- Asked for a *count* instead of a *total*? Swap `sum(...)` for `len(...)`.
- Asked for the smallest instead of the largest? Swap `max` for `min`. Everything else stays.

### Traps that cost marks

- `range(101, 110)` stops at **109**. To include 110 you need `range(101, 111)`.
- `lst.remove('x')` deletes the *first* match only, and crashes if `'x'` isn't there. Check with `if 'x' in lst:` first.
- `lst.reverse()` returns `None`; it changes the list in place. `print(lst.reverse())` prints `None`, which looks like a bug but isn't.
- `d['missing']` crashes with a `KeyError`. `d.get('missing')` quietly returns `None`.
- Dictionary keys must be unique. Assigning the same key twice silently overwrites it, with no warning.
