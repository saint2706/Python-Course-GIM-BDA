```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.model_selection import train_test_split
from scipy.stats import skew

# Generate non-linear synthetic data example: Logarithmic relationship
#np.random.seed(42)
advertising_spend = np.random.uniform(1, 1000, 1000)
sales = 30 + 10 * np.log(advertising_spend) + np.random.normal(0, 2, 1000)

data = pd.DataFrame({'Advertising Spend': advertising_spend, 'Sales': sales})

data.to_csv("Advertising Data")

# Scatter plot
sns.scatterplot(x='Advertising Spend', y='Sales', data=data)
plt.title("Scatter Plot of Original Data (Log Relationship)")
plt.show()

# Split data
X_train, X_test, y_train, y_test = train_test_split(data[['Advertising Spend']], data['Sales'], test_size=0.2, random_state=42)

# Initial Linear Regression Model without transformation
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Initial Performance Metrics
print("Initial Linear Model (No Transformation)")
print(f"R-squared: {r2_score(y_test, y_pred):.4f}")
print(f"RMSE: {mean_squared_error(y_test, y_pred, squared=False):.4f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred):.4f}")

# Apply Log Transformation to X
data['Advertising Spend Transformed'] = np.log(data['Advertising Spend'])


X_train, X_test, y_train, y_test = train_test_split(data[['Advertising Spend Transformed']], data['Sales'], test_size=0.2, random_state=42)

# Revised Linear Model after Log Transformation
model.fit(X_train, y_train)
y_pred_transformed = model.predict(X_test)

# Transformed Model Performance Metrics
print("\nRevised Linear Model (After Log Transformation)")
print(f"R-squared: {r2_score(y_test, y_pred_transformed):.4f}")
print(f"RMSE: {mean_squared_error(y_test, y_pred_transformed, squared=False):.4f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred_transformed):.4f}")

# Visualize the transformed data for linearity
sns.scatterplot(x='Advertising Spend Transformed', y='Sales', data=data)
plt.title("Scatter Plot after Log Transformation (Improved Linearity)")
plt.show()
```
