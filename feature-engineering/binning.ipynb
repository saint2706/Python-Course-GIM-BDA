```python
#import pandas and data set
import pandas as pd
import numpy as np
df = pd.read_csv('student_bucketing.csv', header=0)
#to print first five rows
df.head()
```

```python
#Creating five bins from markw of students
df['bucket']=pd.cut(df['marks'],5,labels=['Poor','Below_average','Average','Above_Average','Excellent'])
df.head(30)
```

```python
# We can check the list of bins 
df['bucket'].unique()
```

```python
df['bins'] = pd.cut(x=df['marks'], bins=[1, 20, 40, 60,80, 100])
```
