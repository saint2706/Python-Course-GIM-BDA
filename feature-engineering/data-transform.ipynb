```python
#importing libraries
import pandas as pd
import numpy as np
from sklearn import preprocessing
```

```python
df = pd.read_csv('data1.csv', header=0)
#to print first five rows
df.head(10)
```

```python
#applying min-max normalisation
df['normalized'] = (df['value'] - df['value'].min()) / (df['value'].max() - df['value'].min())
#check the data contents
df.head()
```

```python
#applying standardization
df['standardized'] = (df['value'] - df['value'].mean()) / df['value'].std()
```

```python
df
```

```python
#applying log transformation
df["log_amount"]=np.log(df["amount"])
```

```python
df
```

```python
#applying base 10 log transformation
df["log10_amount"]=np.log10(df["amount"])
```

```python
df
```

```python
df["sqrt_amount"]=np.sqrt(df["amount"])
```

```python
df = pd.read_csv('data1.csv', header=0)
#to print first five rows
df.head()
```
